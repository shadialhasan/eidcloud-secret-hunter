# 🛡️ eidcloud-secret-hunter: Quickstart Guide

Welcome to **eidcloud-secret-hunter**, a high-velocity secrets and credential scanner built with pure PHP 8.2+ (zero vendor dependencies).

### Features Demonstrated:
1. **25+ High-Precision Signatures**: Detect OpenAI, AWS, Google Cloud, GitHub, Stripe, SSH keys, Database URLs, etc.
2. **Shannon Entropy Analysis**: Flags high-entropy random strings while preventing false positives.
3. **Automatic Masking**: Redacts sensitive payloads (e.g. `sk-proj-****AbCd`) in terminal and CI logs.
4. **CI/CD Blocker Exit Codes**: Exit code `1` on critical/high leaks, `0` on clean state.
